```python
import pandas as pd

# Lê o JSON definindo a codificação correta para caracteres em português
df = pd.read_json('D.SDA.PDA.001.CON.202305.json', encoding='latin1')

# Exibe o tamanho do arquivo
print(f"Formato do arquivo: {df.shape}")

# Exibe as primeiras linhas
df.head()
```

```python
df.info()
```

```python
import pandas as pd

# Expande o conteúdo da coluna 'nodes' em colunas individuais
df_exp = pd.json_normalize(df["nodes"])

# Exibe o novo total de linhas e colunas
print(f"Novo formato do DataFrame: {df_exp.shape}")

# Lista todas as novas colunas encontradas
print("\nColunas encontradas:")
print(df_exp.columns.tolist())

# Visualiza as primeiras linhas da tabela tratada
df_exp.head()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

# 1. Distribuição de Concessões por Tipo de Despacho (Corrigido)
plt.figure(figsize=(12, 8))
ordem_despacho = df_exp['node.Despacho'].value_counts().head(15).index

sns.countplot(
    data=df_exp[df_exp['node.Despacho'].isin(ordem_despacho)],
    y='node.Despacho',
    order=ordem_despacho,
    hue='node.Despacho',
    palette='Blues_r',
    legend=False,
)

plt.title(
    'Top 15 Tipos de Despacho na Concessão de Benefícios',
    fontsize=14,
    pad=15,
)
plt.xlabel('Quantidade de Benefícios', fontsize=12)
plt.ylabel('Tipo de Despacho', fontsize=12)
plt.tight_layout()
plt.show()
```

```python
# Tabela com contagem absoluta e percentual
tabela_despacho = df_exp['node.Despacho'].value_counts().reset_index()
tabela_despacho.columns = ['Tipo de Despacho', 'Total']

# Calcula a porcentagem
tabela_despacho['Porcentagem (%)'] = (
    tabela_despacho['Total'] / len(df_exp)
) * 100

# Exibe os 10 mais relevantes
tabela_despacho.head(10)
```

```python
# 1. Filtra apenas os casos judiciais
df_judicial = df_exp[
    df_exp['node.Despacho'] == 'Concessao Decorrente de Acao Judicial'
]

# 2. Top 10 Benefícios Concedidos na Justiça
print('--- TOP 10 BENEFÍCIOS JUDICIAIS ---')
top_especies_jud = df_judicial['node.Espécie'].value_counts().head(10)
print(top_especies_jud)

print('\n--- TOP 10 CIDs (DIAGNÓSTICOS) NA JUSTIÇA ---')
# 3. Top 10 Diagnósticos (CIDs) mais comuns na Justiça (excluindo zerados)
top_cids_jud = (
    df_judicial[df_judicial['node.CID'] != 'Zerados']['node.CID']
    .value_counts()
    .head(10)
)
print(top_cids_jud)
```

```python
from datetime import datetime

# Converte data de nascimento para formato datetime
df_exp['dt_nasc'] = pd.to_datetime(
    df_exp['node.Dt Nascimento'], format='%d/%m/%Y', errors='coerce'
)

# Considera o ano da competência (2023) para cálculo da idade aproximada
df_exp['idade'] = 2023 - df_exp['dt_nasc'].dt.year

# Gráfico de Distribuição de Idade por Via de Concessão
plt.figure(figsize=(10, 5))
sns.boxplot(
    data=df_exp[
        df_exp['node.Despacho'].isin(
            ['Concessao Normal', 'Concessao Decorrente de Acao Judicial']
        )
    ],
    x='node.Despacho',
    y='idade',
    palette='Set2',
)

plt.title('Comparativo de Idade: Concessão Administrativa vs. Judicial')
plt.xlabel('Via de Concessão')
plt.ylabel('Idade (anos)')
plt.tight_layout()
plt.show()
```
